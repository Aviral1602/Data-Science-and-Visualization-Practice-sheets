### Topic 8: Scope — Local vs Global Variables

#### Question 8.1

````python
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(-5))
# print(val) would cause a NameError because val is local to the function.
````

#### Question 8.2

````python
counter = 0

def show_counter():
    print(counter)

show_counter()
````

#### Question 8.3

````python
counter = 0

def increment_counter():
    counter = counter + 1

# Calling increment_counter() raises UnboundLocalError because counter is treated as local.
````

#### Question 8.4

````python
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
````
