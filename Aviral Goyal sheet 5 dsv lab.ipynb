```python
np.random.seed(35)
marks=np.random.randint(20,101,(12,4))
avg=np.mean(marks,axis=1)
testmean=np.mean(marks,axis=0)
teststd=np.std(marks,axis=0)
passed=np.all(marks>=40,axis=1)
highavg=np.where(avg>=75)[0]+1
print(marks)
print(avg)
print(testmean)
print(teststd)
print(np.sum(passed))
print(highavg)
print(np.argmax(teststd)+1)
```

Q37

```python
quantity=np.array([[12,18,20,15],[16,22,17,19],[14,20,24,18],[21,16,19,23],[18,25,22,20],[20,19,21,24]])
price=np.array([150,240,90,320])
daily=np.sum(quantity*price,axis=1)
product=np.sum(quantity,axis=0)
revenue=np.sum(quantity*price,axis=0)
newprice=np.where(product>100,price*0.9,price)
newrevenue=np.sum(quantity*newprice,axis=0)
print(daily)
print(revenue)
print(np.argmax(daily)+1)
print(np.argmax(product)+1)
print(newprice)
print(np.sum(newrevenue))
```

Q38

```python
np.random.seed(38)
data=np.random.normal(50,12,100)
clean=np.where(data<30,30,np.where(data>70,70,data))
print(np.sum((data<30)|(data>70)))
print(np.mean(data),np.mean(clean))
print(np.var(data),np.var(clean))
print(np.std(data),np.std(clean))
print(np.min(data),np.min(clean))
print(np.max(data),np.max(clean))
```

Q39

```python
branch_a=pd.Series({'Q1':120,'Q2':145,'Q4':180})
branch_b=pd.Series({'Q2':110,'Q3':135,'Q4':155})
branch_c=pd.Series({'Q1':95,'Q3':125,'Q4':140})
result=branch_a+branch_b+branch_c
common=branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
zero_result=branch_a.add(branch_b,fill_value=0).add(branch_c,fill_value=0)
print(result)
print(common)
print(result[common])
print(zero_result)
```

```python
scores={'R101':78,'R102':65,'R104':92,'R105':48}
required=['R105','R103','R101','R106','R104','R102']
s=pd.Series(scores,index=required)
valid=s[s.notnull() & (s>=70)]
result=valid.to_dict()
new=pd.Series(result)
print(s)
print(s[s.isnull()])
print(valid)
print(result)
print(new)
print(valid.equals(new))
```
